## EDA

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# =====================================================
# HANDLE NULL VALUES
# =====================================================
df["description"] = df["description"].fillna("No description")
df["homepage"] = df["homepage"].fillna("No homepage")
df["language"] = df["language"].fillna("Unknown")
df["license"] = df["license"].fillna("No license")
df["topics"] = df["topics"].fillna("No topics")

# Convert dates
date_cols = ["created_at", "updated_at", "pushed_at"]
for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors="coerce")

sns.set_theme(style="whitegrid")

# =====================================================
# 1. TOP LANGUAGES
# =====================================================
top_lang = df["language"].value_counts().head(10)

plt.figure(figsize=(12, 6))
sns.barplot(
    x=top_lang.index,
    y=top_lang.values,
    hue=top_lang.index,
    palette="Set2",
    legend=False
)
plt.title("Top 10 Programming Languages", fontsize=16)
plt.xlabel("Language")
plt.ylabel("Repository Count")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# =====================================================
# 2. STARS DISTRIBUTION
# =====================================================
plt.figure(figsize=(10, 6))
sns.histplot(df["stars"], bins=30, kde=True, color="purple")
plt.title("Stars Distribution", fontsize=16)
plt.xlabel("Stars")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()


# =====================================================
# 3. FORKS VS WATCHERS
# =====================================================
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="forks",
    y="watchers",
    hue="language",
    palette="tab20",
    alpha=0.8,
    s=70
)
plt.title("Forks vs Watchers", fontsize=16)
plt.tight_layout()
plt.show()


# =====================================================
# 4. TOP 10 REPOSITORIES BY STARS
# =====================================================
top_stars = df.nlargest(10, "stars")

plt.figure(figsize=(12, 6))
sns.barplot(
    data=top_stars,
    x="name",
    y="stars",
    hue="name",
    palette="viridis",
    legend=False
)
plt.title("Top 10 Repositories by Stars", fontsize=16)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()


# =====================================================
# 5. LICENSE DISTRIBUTION
# =====================================================
top_license = df["license"].value_counts().head(8)

plt.figure(figsize=(10, 10))
plt.pie(
    top_license.values,
    labels=top_license.index,
    autopct="%1.1f%%",
    colors=sns.color_palette("pastel")
)
plt.title("License Distribution", fontsize=16)
plt.show()


# =====================================================
# 6. REPOSITORY AGE DISTRIBUTION
# =====================================================
plt.figure(figsize=(10, 6))
sns.histplot(df["repo_age_days"], bins=30, kde=True, color="orange")
plt.title("Repository Age Distribution", fontsize=16)
plt.xlabel("Age (Days)")
plt.tight_layout()
plt.show()


# =====================================================
# 7. SIZE DISTRIBUTION BY LANGUAGE
# =====================================================
top5_lang = df["language"].value_counts().head(5).index
filtered = df[df["language"].isin(top5_lang)]

plt.figure(figsize=(12, 6))
sns.boxplot(
    data=filtered,
    x="language",
    y="size_kb",
    hue="language",
    palette="Set3",
    legend=False
)
plt.title("Repository Size by Language", fontsize=16)
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()


# =====================================================
# 8. STARS PER DAY BY LANGUAGE
# =====================================================
plt.figure(figsize=(12, 6))
sns.boxplot(
    data=filtered,
    x="language",
    y="stars_per_day",
    hue="language",
    palette="Spectral",
    legend=False
)
plt.title("Stars Per Day by Language", fontsize=16)
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()


# =====================================================
# 9. OFFICIAL DISCORD COUNT
# =====================================================
plt.figure(figsize=(6, 6))
sns.countplot(
    data=df,
    x="is_official_discord",
    hue="is_official_discord",
    palette="coolwarm",
    legend=False
)
plt.title("Repositories with Official Discord", fontsize=16)
plt.xlabel("Has Official Discord")
plt.ylabel("Count")
plt.tight_layout()
plt.show()


# =====================================================
# 10. FORKED VS ORIGINAL REPOSITORIES
# =====================================================
plt.figure(figsize=(6, 6))
sns.countplot(
    data=df,
    x="is_fork",
    hue="is_fork",
    palette="Accent",
    legend=False
)
plt.title("Forked vs Original Repositories", fontsize=16)
plt.tight_layout()
plt.show()


# =====================================================
# 11. ARCHIVED REPOSITORIES
# =====================================================
plt.figure(figsize=(6, 6))
sns.countplot(
    data=df,
    x="archived",
    hue="archived",
    palette="Dark2",
    legend=False
)
plt.title("Archived Repositories", fontsize=16)
plt.tight_layout()
plt.show()


# =====================================================
# 12. OPEN ISSUES VS STARS
# =====================================================
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="open_issues",
    y="stars",
    hue="owner_type",
    palette="Paired",
    alpha=0.8,
    s=70
)
plt.title("Open Issues vs Stars", fontsize=16)
plt.tight_layout()
plt.show()


# =====================================================
# 13. REPOSITORIES CREATED PER YEAR
# =====================================================
df["created_year"] = df["created_at"].dt.year
year_count = df["created_year"].value_counts().sort_index()

plt.figure(figsize=(12, 6))
sns.lineplot(
    x=year_count.index,
    y=year_count.values,
    marker="o",
    linewidth=2.5,
    color="crimson"
)
plt.title("Repositories Created Per Year", fontsize=16)
plt.xlabel("Year")
plt.ylabel("Repository Count")
plt.tight_layout()
plt.show()


# =====================================================
# 14. CORRELATION HEATMAP
# =====================================================
numeric_cols = [
    "stars", "forks", "watchers", "open_issues",
    "stars_per_day", "repo_age_days",
    "days_since_last_push", "size_kb"
]

plt.figure(figsize=(12, 8))
corr = df[numeric_cols].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="Spectral",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap", fontsize=16)
plt.tight_layout()
plt.show()


# =====================================================
# 15. DAYS SINCE LAST PUSH DISTRIBUTION
# =====================================================
plt.figure(figsize=(10, 6))
sns.histplot(
    df["days_since_last_push"],
    bins=30,
    kde=True,
    color="teal"
)
plt.title("Days Since Last Push Distribution", fontsize=16)
plt.xlabel("Days Since Last Push")
plt.tight_layout()
plt.show()


# =====================================================
# 16. WATCHERS BY OWNER TYPE
# =====================================================
plt.figure(figsize=(8, 6))
sns.boxplot(
    data=df,
    x="owner_type",
    y="watchers",
    hue="owner_type",
    palette="rainbow",
    legend=False
)
plt.title("Watchers by Owner Type", fontsize=16)
plt.tight_layout()
plt.show()


# =====================================================
# 17. TOPICS LENGTH ANALYSIS
# =====================================================
df["topic_count"] = df["topics"].astype(str).apply( lambda x: len(x.split(",")) if x != "No topics" else 0 )

plt.figure(figsize=(10, 6))
sns.histplot(df["topic_count"], bins=15, kde=True, color="magenta")
plt.title("Number of Topics per Repository", fontsize=16)
plt.xlabel("Topic Count")
plt.tight_layout()
plt.show()


# =====================================================
# 18. TOP OWNERS BY TOTAL STARS
# =====================================================
top_owners = (
    df.groupby("owner")["stars"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)

plt.figure(figsize=(12, 6))
sns.barplot(
    data=top_owners,
    x="owner",
    y="stars",
    hue="owner",
    palette="cubehelix",
    legend=False
)
plt.title("Top Owners by Total Stars", fontsize=16)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
