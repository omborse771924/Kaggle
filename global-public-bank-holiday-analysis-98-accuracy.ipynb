## Loading Dataset

```python
# ============================================================
# GLOBAL PUBLIC HOLIDAYS 2020-2026
# COMPLETE DATA LOADING + EDA
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. LOAD DATASET
# ============================================================

file_path = "/kaggle/input/datasets/mubashirulhassan00/global-public-holidays-and-bank-holidays-20202026/global_public_holidays_2020_2026.csv"

df = pd.read_csv(file_path)

print("=" * 60)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 60)

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

# ============================================================
# 2. DISPLAY BASIC DATA
# ============================================================

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())

print("\nColumn Names:")
print(df.columns.tolist())

# ============================================================
# 3. DATA TYPES
# ============================================================

print("\n" + "=" * 60)
print("DATA TYPES")
print("=" * 60)

display(df.dtypes.to_frame(name="Data Type"))

# ============================================================
# 4. DATASET INFORMATION
# ============================================================

print("\n" + "=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

df.info()

# ============================================================
# 5. STATISTICAL SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("STATISTICAL SUMMARY")
print("=" * 60)

display(df.describe(include="all").T)

# ============================================================
# 6. MISSING VALUES
# ============================================================

print("\n" + "=" * 60)
print("MISSING VALUE ANALYSIS")
print("=" * 60)

missing_data = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

display(missing_data)

# ============================================================
# 7. DUPLICATE VALUES
# ============================================================

print("\n" + "=" * 60)
print("DUPLICATE ANALYSIS")
print("=" * 60)

duplicate_count = df.duplicated().sum()

print("Duplicate Rows:", duplicate_count)

# ============================================================
# 8. UNIQUE VALUES
# ============================================================

print("\n" + "=" * 60)
print("UNIQUE VALUES")
print("=" * 60)

for column in df.columns:
    print(f"{column}: {df[column].nunique()} unique values")

# ============================================================
# 9. IDENTIFY DATE COLUMN AUTOMATICALLY
# ============================================================

print("\n" + "=" * 60)
print("DATE COLUMN DETECTION")
print("=" * 60)

date_column = None

# First try columns whose names indicate a date
possible_date_columns = [
    col for col in df.columns
    if any(word in col.lower() for word in
           ["date", "day", "datetime"])
]

# Try converting candidate columns
for col in possible_date_columns:

    converted = pd.to_datetime(
        df[col],
        errors="coerce"
    )

    valid_percentage = converted.notna().mean()

    if valid_percentage > 0.5:
        date_column = col
        df[col] = converted
        break

# If no date column was found, try every column
if date_column is None:

    for col in df.columns:

        converted = pd.to_datetime(
            df[col],
            errors="coerce"
        )

        valid_percentage = converted.notna().mean()

        if valid_percentage > 0.5:
            date_column = col
            df[col] = converted
            break

# ============================================================
# 10. CHECK DATE COLUMN
# ============================================================

if date_column is not None:

    print("Date column detected:", date_column)

    print("\nMinimum Date:", df[date_column].min())
    print("Maximum Date:", df[date_column].max())

else:

    print("WARNING: No suitable date column was detected.")
    print("Date-based EDA will be skipped.")

# ============================================================
# 11. CATEGORICAL DATA ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("CATEGORICAL DATA ANALYSIS")
print("=" * 60)

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

for column in categorical_columns:

    print(f"\nTop values for: {column}")

    display(
        df[column]
        .value_counts(dropna=False)
        .head(15)
        .to_frame("Count")
    )

# ============================================================
# 12. DATE BASED EDA
# ============================================================

if date_column is not None:

    # Remove invalid dates only for date analysis
    date_df = df[df[date_column].notna()].copy()

    if len(date_df) > 0:

        # ====================================================
        # YEAR-WISE ANALYSIS
        # ====================================================

        date_df["Year"] = date_df[date_column].dt.year

        yearly_count = (
            date_df["Year"]
            .value_counts()
            .sort_index()
        )

        print("\n" + "=" * 60)
        print("HOLIDAYS BY YEAR")
        print("=" * 60)

        display(
            yearly_count
            .to_frame("Number of Holidays")
        )

        # Plot only if data exists
        if not yearly_count.empty:

            plt.figure(figsize=(10, 5))

            plt.bar(
                yearly_count.index.astype(str),
                yearly_count.values
            )

            plt.title("Number of Holidays by Year")
            plt.xlabel("Year")
            plt.ylabel("Number of Holidays")
            plt.xticks(rotation=45)

            plt.tight_layout()
            plt.show()

        # ====================================================
        # MONTH-WISE ANALYSIS
        # ====================================================

        date_df["Month"] = date_df[date_column].dt.month

        monthly_count = (
            date_df["Month"]
            .value_counts()
            .sort_index()
        )

        print("\n" + "=" * 60)
        print("HOLIDAYS BY MONTH")
        print("=" * 60)

        display(
            monthly_count
            .to_frame("Number of Holidays")
        )

        if not monthly_count.empty:

            plt.figure(figsize=(10, 5))

            plt.bar(
                monthly_count.index.astype(str),
                monthly_count.values
            )

            plt.title("Number of Holidays by Month")
            plt.xlabel("Month")
            plt.ylabel("Number of Holidays")

            plt.tight_layout()
            plt.show()

        # ====================================================
        # DAY OF WEEK ANALYSIS
        # ====================================================

        date_df["Day"] = date_df[date_column].dt.day_name()

        day_order = [
            "Monday",
            "Tuesday",
            "Wednesday",
            "Thursday",
            "Friday",
            "Saturday",
            "Sunday"
        ]

        day_count = (
            date_df["Day"]
            .value_counts()
            .reindex(day_order, fill_value=0)
        )

        print("\n" + "=" * 60)
        print("HOLIDAYS BY DAY OF WEEK")
        print("=" * 60)

        display(
            day_count
            .to_frame("Number of Holidays")
        )

        plt.figure(figsize=(10, 5))

        plt.bar(
            day_count.index,
            day_count.values
        )

        plt.title("Holidays by Day of Week")
        plt.xlabel("Day")
        plt.ylabel("Number of Holidays")
        plt.xticks(rotation=45)

        plt.tight_layout()
        plt.show()

        # ====================================================
        # QUARTER-WISE ANALYSIS
        # ====================================================

        date_df["Quarter"] = date_df[date_column].dt.quarter

        quarter_count = (
            date_df["Quarter"]
            .value_counts()
            .sort_index()
        )

        print("\n" + "=" * 60)
        print("HOLIDAYS BY QUARTER")
        print("=" * 60)

        display(
            quarter_count
            .to_frame("Number of Holidays")
        )

        if not quarter_count.empty:

            plt.figure(figsize=(8, 5))

            plt.bar(
                ["Q1", "Q2", "Q3", "Q4"][:len(quarter_count)],
                quarter_count.values
            )

            plt.title("Number of Holidays by Quarter")
            plt.xlabel("Quarter")
            plt.ylabel("Number of Holidays")

            plt.tight_layout()
            plt.show()

# ============================================================
# 13. NUMERIC COLUMN ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("NUMERIC COLUMN ANALYSIS")
print("=" * 60)

numeric_columns = df.select_dtypes(
    include=np.number
).columns

if len(numeric_columns) > 0:

    for column in numeric_columns:

        print(f"\nColumn: {column}")

        print("Minimum :", df[column].min())
        print("Maximum :", df[column].max())
        print("Mean    :", df[column].mean())
        print("Median  :", df[column].median())

else:

    print("No numeric columns found.")

# ============================================================
# 14. OUTLIER CHECK FOR NUMERIC COLUMNS
# ============================================================

if len(numeric_columns) > 0:

    print("\n" + "=" * 60)
    print("OUTLIER ANALYSIS")
    print("=" * 60)

    for column in numeric_columns:

        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)

        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = df[
            (df[column] < lower_bound) |
            (df[column] > upper_bound)
        ]

        print(
            f"{column}: {len(outliers)} possible outliers"
        )

# ============================================================
# 15. CORRELATION ANALYSIS
# ============================================================

if len(numeric_columns) >= 2:

    print("\n" + "=" * 60)
    print("CORRELATION MATRIX")
    print("=" * 60)

    correlation = df[numeric_columns].corr()

    display(correlation)

    plt.figure(figsize=(10, 7))

    plt.imshow(
        correlation,
        aspect="auto"
    )

    plt.colorbar()

    plt.xticks(
        range(len(correlation.columns)),
        correlation.columns,
        rotation=90
    )

    plt.yticks(
        range(len(correlation.columns)),
        correlation.columns
    )

    plt.title("Correlation Matrix")

    plt.tight_layout()
    plt.show()

# ============================================================
# 16. FINAL DATASET SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("FINAL EDA SUMMARY")
print("=" * 60)

print("Total Rows             :", df.shape[0])
print("Total Columns          :", df.shape[1])
print("Duplicate Rows         :", df.duplicated().sum())
print("Total Missing Values   :", df.isnull().sum().sum())

if date_column is not None:

    valid_dates = df[date_column].notna().sum()

    print("Valid Date Records     :", valid_dates)
    print("Invalid/Missing Dates  :", len(df) - valid_dates)
    print("Date Column            :", date_column)

print("\nEDA COMPLETED SUCCESSFULLY!")
```

## Features Engineeing

```python
# ============================================================
# FEATURE ENGINEERING
# Global Public Holidays Dataset
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. COPY DATASET
# ------------------------------------------------------------

df_fe = df.copy()

print("Original Shape:", df_fe.shape)

# ------------------------------------------------------------
# 2. FIND DATE COLUMN
# ------------------------------------------------------------

date_column = None

for col in df_fe.columns:

    if "date" in col.lower():

        converted = pd.to_datetime(
            df_fe[col],
            errors="coerce"
        )

        if converted.notna().sum() > 0:
            date_column = col
            df_fe[col] = converted
            break

print("Date Column:", date_column)

# ------------------------------------------------------------
# 3. DATE FEATURES
# ------------------------------------------------------------

if date_column is not None:

    df_fe["year"] = df_fe[date_column].dt.year

    df_fe["month"] = df_fe[date_column].dt.month

    df_fe["day"] = df_fe[date_column].dt.day

    df_fe["day_of_week"] = df_fe[date_column].dt.dayofweek

    df_fe["week_of_year"] = (
        df_fe[date_column]
        .dt.isocalendar()
        .week
        .astype(int)
    )

    df_fe["quarter"] = df_fe[date_column].dt.quarter

    # Weekend = 1, Weekday = 0
    df_fe["is_weekend"] = (
        df_fe["day_of_week"] >= 5
    ).astype(int)

# ------------------------------------------------------------
# 4. MONTH NAME
# ------------------------------------------------------------

if date_column is not None:

    df_fe["month_name"] = (
        df_fe[date_column]
        .dt.month_name()
    )

# ------------------------------------------------------------
# 5. DAY NAME
# ------------------------------------------------------------

if date_column is not None:

    df_fe["day_name"] = (
        df_fe[date_column]
        .dt.day_name()
    )

# ------------------------------------------------------------
# 6. CYCLICAL MONTH FEATURES
# ------------------------------------------------------------

if "month" in df_fe.columns:

    df_fe["month_sin"] = np.sin(
        2 * np.pi * df_fe["month"] / 12
    )

    df_fe["month_cos"] = np.cos(
        2 * np.pi * df_fe["month"] / 12
    )

# ------------------------------------------------------------
# 7. CYCLICAL DAY-OF-WEEK FEATURES
# ------------------------------------------------------------

if "day_of_week" in df_fe.columns:

    df_fe["day_sin"] = np.sin(
        2 * np.pi * df_fe["day_of_week"] / 7
    )

    df_fe["day_cos"] = np.cos(
        2 * np.pi * df_fe["day_of_week"] / 7
    )

# ------------------------------------------------------------
# 8. COUNTRY FREQUENCY FEATURE
# ------------------------------------------------------------

country_column = None

for col in df_fe.columns:

    if "country" in col.lower():

        country_column = col
        break

if country_column is not None:

    country_counts = (
        df_fe[country_column]
        .value_counts()
    )

    df_fe["country_holiday_count"] = (
        df_fe[country_column]
        .map(country_counts)
    )

# ------------------------------------------------------------
# 9. HOLIDAY TYPE FREQUENCY
# ------------------------------------------------------------

type_column = None

for col in df_fe.columns:

    if "type" in col.lower():

        type_column = col
        break

if type_column is not None:

    type_counts = (
        df_fe[type_column]
        .value_counts()
    )

    df_fe["holiday_type_count"] = (
        df_fe[type_column]
        .map(type_counts)
    )

# ------------------------------------------------------------
# 10. HOLIDAY NAME FEATURES
# ------------------------------------------------------------

name_column = None

for col in df_fe.columns:

    if "name" in col.lower():

        if df_fe[col].dtype == "object":

            name_column = col
            break

if name_column is not None:

    # Length of holiday name
    df_fe["holiday_name_length"] = (
        df_fe[name_column]
        .fillna("")
        .astype(str)
        .str.len()
    )

    # Number of words
    df_fe["holiday_name_words"] = (
        df_fe[name_column]
        .fillna("")
        .astype(str)
        .str.split()
        .str.len()
    )

# ------------------------------------------------------------
# 11. DISPLAY FEATURE ENGINEERED DATA
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FEATURE ENGINEERING COMPLETED")
print("=" * 60)

print("Original Shape :", df.shape)
print("New Shape      :", df_fe.shape)

print("\nNew Features:")

new_features = [
    col for col in df_fe.columns
    if col not in df.columns
]

for col in new_features:
    print("-", col)

# ------------------------------------------------------------
# 12. DISPLAY DATA
# ------------------------------------------------------------

print("\nFeature Engineered Dataset:")

display(df_fe.head())

# ------------------------------------------------------------
# 13. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\nMissing Values:")

display(
    df_fe.isnull()
    .sum()
    .to_frame("Missing Values")
)

# ------------------------------------------------------------
# 14. SAVE DATASET
# ------------------------------------------------------------

df_fe.to_csv(
    "/kaggle/working/feature_engineered_holidays.csv",
    index=False
)

print(
    "\nSaved as: "
    "/kaggle/working/feature_engineered_holidays.csv"
)
```

```python
# ============================================================
# TRAIN / TEST + ACCURACY + ROC-AUC
# TARGET = is_global
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_curve,
    roc_auc_score,
    ConfusionMatrixDisplay
)

# ============================================================
# 1. COPY FEATURE ENGINEERED DATA
# ============================================================

data = df_fe.copy()

print("Dataset Shape:", data.shape)

# ============================================================
# 2. SET TARGET
# ============================================================

target = "is_global"

print("Target:", target)

print("\nTarget Distribution:")
print(data[target].value_counts())

# ============================================================
# 3. REMOVE TARGET FROM FEATURES
# ============================================================

X = data.drop(columns=[target])
y = data[target]

# ============================================================
# 4. REMOVE DATE COLUMN
# ============================================================

# Raw date is not needed because we already extracted
# year, month, day, week, quarter etc.

if "date" in X.columns:
    X = X.drop(columns=["date"])

# ============================================================
# 5. REMOVE HIGH-CARDINALITY / TEXT COLUMNS
# ============================================================

columns_to_drop = [
    "country_name",
    "country_code",
    "local_name",
    "english_name",
    "applicable_regions",
    "month_name",
    "day_name"
]

columns_to_drop = [
    col for col in columns_to_drop
    if col in X.columns
]

X = X.drop(columns=columns_to_drop)

# ============================================================
# 6. HANDLE CATEGORICAL COLUMNS
# ============================================================

categorical_columns = X.select_dtypes(
    include=["object", "category"]
).columns

print("\nCategorical Columns:")
print(categorical_columns.tolist())

X = pd.get_dummies(
    X,
    columns=categorical_columns,
    drop_first=True
)

# ============================================================
# 7. HANDLE MISSING / INFINITE VALUES
# ============================================================

X = X.replace(
    [np.inf, -np.inf],
    np.nan
)

X = X.fillna(0)

# ============================================================
# 8. TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n" + "=" * 60)
print("TRAIN / TEST SPLIT")
print("=" * 60)

print("X Train:", X_train.shape)
print("X Test :", X_test.shape)
print("Y Train:", y_train.shape)
print("Y Test :", y_test.shape)

# ============================================================
# 9. RANDOM FOREST CLASSIFIER
# ============================================================

model = RandomForestClassifier(
    n_estimators=200,
    max_depth=15,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

model.fit(
    X_train,
    y_train
)

# ============================================================
# 10. PREDICTION
# ============================================================

y_pred = model.predict(X_test)

y_prob = model.predict_proba(X_test)[:, 1]

# ============================================================
# 11. ACCURACY
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\n" + "=" * 60)
print("ACCURACY")
print("=" * 60)

print(
    f"Accuracy: {accuracy * 100:.2f}%"
)

# ============================================================
# 12. CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y_test,
        y_pred
    )
)

# ============================================================
# 13. CONFUSION MATRIX
# ============================================================

print("\n" + "=" * 60)
print("CONFUSION MATRIX")
print("=" * 60)

cm = confusion_matrix(
    y_test,
    y_pred
)

print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Not Global", "Global"]
)

disp.plot()

plt.title("Confusion Matrix")

plt.tight_layout()
plt.show()

# ============================================================
# 14. ROC-AUC SCORE
# ============================================================

roc_auc = roc_auc_score(
    y_test,
    y_prob
)

print("\n" + "=" * 60)
print("ROC-AUC")
print("=" * 60)

print(
    f"ROC-AUC Score: {roc_auc:.4f}"
)

# ============================================================
# 15. ROC CURVE
# ============================================================

fpr, tpr, thresholds = roc_curve(
    y_test,
    y_prob
)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label=f"Random Forest (AUC = {roc_auc:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Guess"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve")

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# ============================================================
# 16. FEATURE IMPORTANCE
# ============================================================

feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print("\n" + "=" * 60)
print("TOP 15 IMPORTANT FEATURES")
print("=" * 60)

display(
    feature_importance.head(15)
)

# ============================================================
# 17. FEATURE IMPORTANCE VISUALIZATION
# ============================================================

top_features = feature_importance.head(15)

plt.figure(figsize=(10, 6))

plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.xlabel("Importance")
plt.ylabel("Feature")

plt.title(
    "Top 15 Feature Importances"
)

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()

# ============================================================
# 18. FINAL RESULTS
# ============================================================

print("\n" + "=" * 60)
print("FINAL MODEL RESULTS")
print("=" * 60)

print(f"Target       : {target}")
print(f"Accuracy     : {accuracy * 100:.2f}%")
print(f"ROC-AUC      : {roc_auc:.4f}")
print(f"Train Samples: {len(X_train)}")
print(f"Test Samples : {len(X_test)}")
```
