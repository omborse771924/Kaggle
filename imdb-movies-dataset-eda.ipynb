```python
df = pd.read_csv('/kaggle/input/datasets/harishyadav0506/imdb-movies-dataset-csv/imdb_movies_dataset.csv.csv')
```

```python
df['parentImdbId'] = df['parentImdbId'].fillna(
    df['parentImdbId'].mean()
)
```

```python
# ============================================================
# HANDLE NULL VALUES
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("NULL VALUES BEFORE HANDLING")
print("=" * 70)

print(df.isnull().sum())


# ============================================================
# 1. ID / CATEGORICAL COLUMNS
# ============================================================

# parentImdbId is an ID, so DO NOT use mean.
df["parentImdbId"] = df["parentImdbId"].fillna("Unknown")


# contentRating is categorical
df["contentRating"] = df["contentRating"].fillna("Unknown")


# ============================================================
# 2. endYear
# ============================================================

# endYear is numeric/year information.
# If missing, it generally means the title has no recorded
# ending year. We use 0 as a placeholder.

df["endYear"] = pd.to_numeric(
    df["endYear"],
    errors="coerce"
)

df["endYear"] = df["endYear"].fillna(0)


# ============================================================
# 3. grossWorldwide
# ============================================================

df["grossWorldwide"] = pd.to_numeric(
    df["grossWorldwide"],
    errors="coerce"
)

df["grossWorldwide"] = df["grossWorldwide"].fillna(
    df["grossWorldwide"].median()
)


# ============================================================
# 4. metascore
# ============================================================

df["metascore"] = pd.to_numeric(
    df["metascore"],
    errors="coerce"
)

df["metascore"] = df["metascore"].fillna(
    df["metascore"].median()
)


# ============================================================
# 5. CHECK NULL VALUES AFTER HANDLING
# ============================================================

print("\n" + "=" * 70)
print("NULL VALUES AFTER HANDLING")
print("=" * 70)

print(df.isnull().sum())


# ============================================================
# 6. TOTAL NULL VALUES
# ============================================================

total_nulls = df.isnull().sum().sum()

print("\nTotal remaining null values:", total_nulls)


# ============================================================
# 7. DISPLAY DATA
# ============================================================

print("\n" + "=" * 70)
print("DATA AFTER NULL HANDLING")
print("=" * 70)

print(df.head())
```

## EDA

```python
# ============================================================
# MOVIE / TV DATASET - VISUALIZATION
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# General visualization settings
# ------------------------------------------------------------

sns.set_theme(style="whitegrid")

# ============================================================
# 1. DATASET OVERVIEW
# ============================================================

print("Dataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)


# ============================================================
# 2. MOVIE / TV TYPE DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="type",
    hue="type",
    palette="Set2",
    legend=False
)

plt.title("Distribution of Movie / TV Types", fontsize=16)
plt.xlabel("Type")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# 3. TOP GENRES
# ============================================================

# Convert genres to strings and split multiple genres
genre_data = (
    df["genres"]
    .dropna()
    .astype(str)
    .str.split(",")
    .explode()
    .str.strip()
)

top_genres = genre_data.value_counts().head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index,
    hue=top_genres.index,
    palette="husl",
    legend=False
)

plt.title("Top 10 Genres", fontsize=16)
plt.xlabel("Number of Titles")
plt.ylabel("Genre")
plt.tight_layout()
plt.show()


# ============================================================
# 4. RELEASE YEAR DISTRIBUTION
# ============================================================

year_data = pd.to_numeric(
    df["startYear"],
    errors="coerce"
).dropna()

plt.figure(figsize=(12, 6))

sns.histplot(
    year_data,
    bins=30,
    kde=True,
    color="purple"
)

plt.title("Distribution of Release Years", fontsize=16)
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.tight_layout()
plt.show()


# ============================================================
# 5. AVERAGE RATING DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    df["averageRating"],
    bins=20,
    kde=True,
    color="orange"
)

plt.title("Distribution of IMDb Ratings", fontsize=16)
plt.xlabel("Average Rating")
plt.ylabel("Number of Titles")
plt.tight_layout()
plt.show()


# ============================================================
# 6. NUMBER OF VOTES DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    np.log1p(df["numVotes"]),
    bins=30,
    kde=True,
    color="green"
)

plt.title("Distribution of Number of Votes (Log Scale)", fontsize=16)
plt.xlabel("Log(Number of Votes + 1)")
plt.ylabel("Number of Titles")
plt.tight_layout()
plt.show()


# ============================================================
# 7. RATING VS NUMBER OF VOTES
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="numVotes",
    y="averageRating",
    hue="type",
    palette="Set1",
    alpha=0.6
)

plt.xscale("log")

plt.title("IMDb Rating vs Number of Votes", fontsize=16)
plt.xlabel("Number of Votes")
plt.ylabel("Average Rating")
plt.tight_layout()
plt.show()


# ============================================================
# 8. RATING BY CONTENT RATING
# ============================================================

top_content = (
    df["contentRating"]
    .fillna("Unknown")
    .value_counts()
    .head(10)
    .index
)

content_data = df[
    df["contentRating"]
    .fillna("Unknown")
    .isin(top_content)
].copy()

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=content_data,
    x="contentRating",
    y="averageRating",
    hue="contentRating",
    palette="Set3",
    legend=False
)

plt.title("IMDb Rating by Content Rating", fontsize=16)
plt.xlabel("Content Rating")
plt.ylabel("Average IMDb Rating")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# 9. TOP COUNTRIES
# ============================================================

country_data = (
    df["countriesOfOrigin"]
    .dropna()
    .astype(str)
    .str.split(",")
    .explode()
    .str.strip()
)

top_countries = country_data.value_counts().head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index,
    hue=top_countries.index,
    palette="cool",
    legend=False
)

plt.title("Top 10 Countries of Origin", fontsize=16)
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.tight_layout()
plt.show()


# ============================================================
# 10. RUNTIME DISTRIBUTION
# ============================================================

runtime = pd.to_numeric(
    df["runtimeMinutes"],
    errors="coerce"
).dropna()

plt.figure(figsize=(10, 6))

sns.histplot(
    runtime,
    bins=30,
    kde=True,
    color="red"
)

plt.title("Distribution of Runtime", fontsize=16)
plt.xlabel("Runtime (Minutes)")
plt.ylabel("Number of Titles")
plt.tight_layout()
plt.show()


# ============================================================
# 11. RUNTIME VS RATING
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="runtimeMinutes",
    y="averageRating",
    hue="type",
    palette="bright",
    alpha=0.6
)

plt.title("Runtime vs IMDb Rating", fontsize=16)
plt.xlabel("Runtime (Minutes)")
plt.ylabel("Average Rating")
plt.tight_layout()
plt.show()


# ============================================================
# 12. BUDGET VS WORLDWIDE GROSS
# ============================================================

budget_data = df[
    (pd.to_numeric(df["budget"], errors="coerce") > 0) &
    (pd.to_numeric(df["grossWorldwide"], errors="coerce") > 0)
].copy()

budget_data["budget"] = pd.to_numeric(
    budget_data["budget"],
    errors="coerce"
)

budget_data["grossWorldwide"] = pd.to_numeric(
    budget_data["grossWorldwide"],
    errors="coerce"
)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=budget_data,
    x="budget",
    y="grossWorldwide",
    hue="type",
    palette="Dark2",
    alpha=0.6
)

plt.xscale("log")
plt.yscale("log")

plt.title("Budget vs Worldwide Gross", fontsize=16)
plt.xlabel("Budget")
plt.ylabel("Worldwide Gross")
plt.tight_layout()
plt.show()


# ============================================================
# 13. ADULT VS NON-ADULT TITLES
# ============================================================

adult_counts = df["isAdult"].value_counts()

plt.figure(figsize=(8, 6))

plt.pie(
    adult_counts.values,
    labels=adult_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=sns.color_palette("Set2")
)

plt.title("Adult vs Non-Adult Titles", fontsize=16)
plt.show()


# ============================================================
# 14. TOP DIRECTORS
# ============================================================

director_data = (
    df["directors"]
    .dropna()
    .astype(str)
    .str.split(",")
    .explode()
    .str.strip()
)

top_directors = director_data.value_counts().head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_directors.values,
    y=top_directors.index,
    hue=top_directors.index,
    palette="Spectral",
    legend=False
)

plt.title("Top 10 Directors", fontsize=16)
plt.xlabel("Number of Titles")
plt.ylabel("Director")
plt.tight_layout()
plt.show()


# ============================================================
# 15. TOP ACTORS / CAST
# ============================================================

cast_data = (
    df["cast"]
    .dropna()
    .astype(str)
    .str.split(",")
    .explode()
    .str.strip()
)

top_cast = cast_data.value_counts().head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_cast.values,
    y=top_cast.index,
    hue=top_cast.index,
    palette="rainbow",
    legend=False
)

plt.title("Top 10 Most Frequent Cast Members", fontsize=16)
plt.xlabel("Number of Titles")
plt.ylabel("Actor / Cast Member")
plt.tight_layout()
plt.show()


# ============================================================
# 16. CORRELATION HEATMAP
# ============================================================

numeric_columns = [
    "startYear",
    "endYear",
    "budget",
    "grossWorldwide",
    "isAdult",
    "runtimeMinutes",
    "averageRating",
    "numVotes",
    "metascore"
]

# Convert columns safely to numeric
numeric_data = df[numeric_columns].apply(
    pd.to_numeric,
    errors="coerce"
)

correlation = numeric_data.corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="Spectral",
    linewidths=0.5
)

plt.title("Correlation Heatmap", fontsize=16)
plt.tight_layout()
plt.show()


# ============================================================
# 17. AVERAGE RATING BY TYPE
# ============================================================

rating_by_type = (
    df.groupby("type")["averageRating"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=rating_by_type.index,
    y=rating_by_type.values,
    hue=rating_by_type.index,
    palette="viridis",
    legend=False
)

plt.title("Average IMDb Rating by Type", fontsize=16)
plt.xlabel("Type")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# 18. TOP 10 TITLES BY RATING
# ============================================================

top_rated = (
    df[["primaryTitle", "averageRating", "numVotes"]]
    .dropna()
    .sort_values("averageRating", ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=top_rated,
    x="averageRating",
    y="primaryTitle",
    hue="primaryTitle",
    palette="magma",
    legend=False
)

plt.title("Top 10 Highest Rated Titles", fontsize=16)
plt.xlabel("Average Rating")
plt.ylabel("Title")
plt.tight_layout()
plt.show()


print("\nVisualization completed successfully! 🎬📊")
```
